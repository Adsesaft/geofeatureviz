```python
import yaml

from geofeatureviz.io import anki_connector
from geofeatureviz_scripts import loader, path_settings
```

# regional_groups.yaml
This is the code I used to create / validate `regional_groups.yaml`. I created the yaml using ChatGPT and first googled every group to confirm that everything is correct. Additionally, I wanted some consistency with my Anki deck regarding the country names. Therefore, I load Ultimate Geography, take the country names and check, that the country names in `regional_groups.yaml` are the same as in Ultimate Geography.

```python
df = anki_connector.deck_to_df(
    "Allgemeinwissen::01 🟦 Geografie 🌍::01.01 Ultimate Geography📍"
)
anki_countries = set(df["Country"])
```

```python
regional_groups = loader.load_regional_groups()

all_countries = []
for group_name, group_info in regional_groups.items():
    core = group_info["core"]
    all_countries.extend(core)

    optional = group_info["optional"]
    all_countries.extend(optional)
json_countries = set(all_countries)

# if this is empty, all json_countries are also in anki_countries
json_countries - anki_countries
```

Furthermore, I want the lists of countries to be alphabetically ordered.

```python
# The lists are sorted in place
[
    c_list.sort()
    for d in regional_groups.values()
    for c_list in d.values()
    if isinstance(c_list, list)
]

file_path = path_settings.regional_groups
with open(file_path, "w", encoding="utf-8") as f:
    yaml.dump(regional_groups, f, sort_keys=False, allow_unicode=True)
```
