```python
from IPython.display import display

from geofeatureviz import preprocessor, svg_handler
from geofeatureviz.data import loader
```

Check the polygon that is created in OrthoMapSVG, which shows the part in longitude/latitude-values that is visible in the orthographic presentation.

```python
countries = preprocessor.prep_dataset("country", "ne", resolution=110)

regional_groups = loader.load_regional_groups()
country_transl = loader.load_country_translations()
```

```python
c = svg_handler.OrthoMapSVG(center=(100, 20), width=500)
g = c.visible_lon_lat

canvas = svg_handler.MapSVG(
    bounds=(-180, -90, 180, 90), width=500, height=250, projection=lambda *x: x
)
canvas.add_gdf(countries, "countries")

elem = canvas.geom_to_svg(g, "a", fill="transparent", stroke="red")
canvas.add(elem)

canvas
```

```python
for lon0 in range(-180, 181, 20):
    c = svg_handler.OrthoMapSVG(center=(lon0, -20), width=200)
    g = c.visible_lon_lat

    canvas = svg_handler.MapSVG(
        bounds=(-180, -90, 180, 90), width=200, height=100, projection=lambda *x: x
    )
    canvas.add_gdf(countries, "countries")

    elem = canvas.geom_to_svg(g, "a", fill="transparent", stroke="red")
    canvas.add(elem)

    display(canvas)
```
